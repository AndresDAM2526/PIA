### Actividad: sistema de ecuaciones

```python
A=np.array([[1,1],[1,-1]])
b=np.array([7,1])
solucion=np.linalg.solve(A,b)
print("A=",solucion[0],"B=",solucion[1])
print("Comprobacion:",A @ solucion)
```

Datos: registro de los precios de un juego a lo largo de 20 meses


```python
precios=np.array([70,70,68,64,63,60,55,52,58,60,65,61,58,56,50,52,54,51,49,45])
print("Registro:",len(precios))
print("Precio máximo:",precios.max())
print("Precio mínimo:",precios.min())
print("Media:",np.mean(precios),"euros")
print("Mediana:",np.median(precios))
print("Desviación típica plobacional:",np.std(precios),"euros")
media_1=np.mean(precios[:10])
media_2=np.mean(precios[10:])
print("Media de los precios(1-10):",media_1)
print("Media de los precios(10-20):",media_2)

```

```python
numero=np.arange(1,len(precios)+1)
plt.figure(figsize=(8,4))
plt.plot(numero,precios,marker="o")
plt.title("Evolución de los precios de un videojuego")
plt.xlabel("Meses")
plt.ylabel("Precios")
plt.grid(True)
plt.show()
```

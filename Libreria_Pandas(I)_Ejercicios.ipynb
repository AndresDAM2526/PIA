- Analizando el dataset ratings.dat, ¿hay algún usuario que no tenga ninguna review? ¿Cuántos tienen menos de 30 reviews?

```python
import numpy as np
import pandas as pd
import os
```

```python
ruta_movies = os.path.join("res", "ml-1m", "movies.dat")
ruta_ratings = os.path.join("res", "ml-1m", "ratings.dat")
ruta_users=os.path.join("res", "ml-1m", "users.dat")
 
movies_dataset = pd.read_csv(ruta_movies, sep='::', index_col=0, engine='python',names=['ID','Nombre','Genero'])
display(movies_dataset.describe())
```

```python
ratings_dataset = pd.read_csv(ruta_ratings, sep='::', engine='python',names=['UserID','MovieID','Rating','TimeStamp'])
display(ratings_dataset.describe())
display(ratings_dataset)
```

```python
users_dataset=pd.read_csv(ruta_users, sep='::', index_col=0,engine='python',names=['UserID','Gender','Age','Occupation','Zip-code'])
users=users_dataset.index
users_ratings=ratings_dataset.groupby('UserID').size()
users_sin_reviews=pd.DataFrame(columns=['UserID'])
print(users)
print("--")
print(users_ratings)


for user in users_dataset.index:
    if user not in users_ratings:
        users_sin_reviews[user]

display(users_sin_reviews)

```

```python
users_ratings=ratings_dataset.groupby('UserID').size()
user_reviews=users_ratings[users_ratings<30]
print(user_reviews)
```
